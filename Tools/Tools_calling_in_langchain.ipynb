```python
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
```

```python
import os

os.environ["GROQ_API_KEY"] = "Enter Your Api Key"
```

**Custom Tool Creation**

```python
@tool
def calculate_percentage(obtained_marks: float,total_marks: float)->float:
  """Calculate percentage from obtained marks and total marks"""
  return(obtained_marks/total_marks)*100
```

```python
print(calculate_percentage.name)
print()
print(calculate_percentage.description)
```

```python
llm = ChatGroq(model = 'openai/gpt-oss-20b',temperature = 0)
```

**Bind a llm with tool**

```python
llm_with_tools = llm.bind_tools(
    [calculate_percentage]
)
```

```python
query = HumanMessage('obtined marks are 560 out of 1000 calculate the percentage')
```

```python
messages = [query]
```

```python
messages
```

**Tool Calling**

```python
result = llm_with_tools.invoke(
    messages
)
print(result)
```

```python
messages.append(result)
messages
```

**Tool execution**

```python
res = calculate_percentage.invoke(result.tool_calls[0])
messages.append(res)
messages
```

```python
cont = llm_with_tools.invoke(messages)
print(cont.content)
```
